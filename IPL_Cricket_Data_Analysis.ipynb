```python
# Step 29: Toss Impact Analysis

df['toss_match_result'] = df.apply(
    lambda row: 'Won Match' if row['toss_winner'] == row['winner']
    else 'Lost Match',
    axis=1
)

df['toss_match_result'].value_counts()
```

```python
toss_impact = df['toss_match_result'].value_counts()

toss_win_percentage = (
    toss_impact['Won Match'] / toss_impact.sum()
) * 100

print(f"Toss winner also won the match: {toss_win_percentage:.2f}%")
```

```python
toss_impact_df = toss_impact.reset_index()

toss_impact_df.columns = ['Result', 'Matches']

toss_impact_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    toss_impact.index,
    toss_impact.values
)

plt.title('Impact of Toss Winning on Match Results')
plt.xlabel('Toss Winner Result')
plt.ylabel('Number of Matches')

plt.tight_layout()
plt.show()
```

```python
toss_percentages = (
    df['toss_match_result']
    .value_counts(normalize=True)
    * 100
)

print(toss_percentages.round(2))
```

### Insight

The Toss Impact Analysis compares the match outcomes of teams that won the toss.

The analysis shows the percentage of matches in which the toss-winning team also won the match. This helps determine whether winning the toss provides a meaningful advantage in IPL matches.

The results suggest that toss outcomes should be evaluated alongside other factors such as team performance, venue, batting first, and bowling conditions.

```python
# Step 31 - Season Analysis

season_df = (
    df.groupby('season')
      .size()
      .reset_index(name='Matches')
)

season_df['Change from Previous Season'] = (
    season_df['Matches'].diff()
)

season_df['Growth %'] = (
    season_df['Matches'].pct_change() * 100
)

season_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    season_df['season'],
    season_df['Matches'],
    marker='o'
)

plt.title('IPL Matches Played by Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

The analysis demonstrates the growth and evolution of the IPL as a major cricket tournament.

```python
# ==========================================
# STEP 32 - FINAL DATA INSIGHTS
# ==========================================

import pandas as pd

# Load datasets
df = pd.read_csv("../data/matches.csv")
deliveries = pd.read_csv("../data/deliveries.csv")

# ------------------------------------------
# 1. Team Performance
# ------------------------------------------

team_wins = df["winner"].dropna().value_counts()

top_team = team_wins.index[0]
top_team_wins = team_wins.iloc[0]


# ------------------------------------------
# 2. Top Batsman
# ------------------------------------------

top_batsmen = (
    deliveries.groupby("batter")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
)

top_batsman = top_batsmen.index[0]
top_batsman_runs = top_batsmen.iloc[0]


# ------------------------------------------
# 3. Top Bowler
# ------------------------------------------

valid_dismissals = [
    "run out",
    "retired hurt",
    "obstructing the field"
]

bowler_wickets = (
    deliveries[
        deliveries["dismissal_kind"].notna()
        & ~deliveries["dismissal_kind"].isin(valid_dismissals)
    ]
    .groupby("bowler")
    .size()
    .sort_values(ascending=False)
)

top_bowler = bowler_wickets.index[0]
top_bowler_wickets = bowler_wickets.iloc[0]


# ------------------------------------------
# 4. Player of the Match
# ------------------------------------------

pom_counts = df["player_of_match"].dropna().value_counts()

top_pom_player = pom_counts.index[0]
top_pom_awards = pom_counts.iloc[0]


# ------------------------------------------
# 5. Toss Impact
# ------------------------------------------

toss_matches = df[
    df["toss_winner"].notna()
    & df["winner"].notna()
]

toss_winner_match_wins = (
    toss_matches["toss_winner"] == toss_matches["winner"]
).sum()

toss_win_percentage = (
    toss_winner_match_wins / len(toss_matches)
) * 100


# ------------------------------------------
# 6. Top Venue
# ------------------------------------------

venue_counts = df["venue"].dropna().value_counts()

top_venue = venue_counts.index[0]
top_venue_matches = venue_counts.iloc[0]


# ------------------------------------------
# 7. Top Season
# ------------------------------------------

season_counts = df["season"].value_counts()

top_season = season_counts.index[0]
top_season_matches = season_counts.iloc[0]


# ==========================================
# FINAL INSIGHTS TABLE
# ==========================================

final_insights = pd.DataFrame({
    "Analysis": [
        "Team Performance",
        "Top Batsman",
        "Top Bowler",
        "Player of the Match",
        "Toss Impact",
        "Top Venue",
        "Top Season"
    ],

    "Key Finding": [
        f"{top_team} ({top_team_wins} wins)",
        f"{top_batsman} ({top_batsman_runs} runs)",
        f"{top_bowler} ({top_bowler_wickets} wickets)",
        f"{top_pom_player} ({top_pom_awards} awards)",
        f"{toss_win_percentage:.2f}% toss winners also won",
        f"{top_venue} ({top_venue_matches} matches)",
        f"{top_season} ({top_season_matches} matches)"
    ]
})

final_insights
```

```python
# ==========================================
# STEP 32 - FINAL DATA INSIGHTS
# ==========================================

import pandas as pd

# Load IPL datasets
df = pd.read_csv("../data/matches.csv")
deliveries = pd.read_csv("../data/deliveries.csv")

# ==========================================
# 1. TEAM PERFORMANCE
# ==========================================

team_wins = df["winner"].dropna().value_counts()

top_team = team_wins.index[0]
top_team_wins = team_wins.iloc[0]


# ==========================================
# 2. TOP BATSMAN
# ==========================================

top_batsmen = (
    deliveries
    .groupby("batter")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
)

top_batsman = top_batsmen.index[0]
top_batsman_runs = top_batsmen.iloc[0]


# ==========================================
# 3. TOP BOWLER
# ==========================================

excluded_dismissals = [
    "run out",
    "retired hurt",
    "obstructing the field"
]

bowler_wickets = (
    deliveries[
        deliveries["dismissal_kind"].notna()
        & ~deliveries["dismissal_kind"].isin(excluded_dismissals)
    ]
    .groupby("bowler")
    .size()
    .sort_values(ascending=False)
)

top_bowler = bowler_wickets.index[0]
top_bowler_wickets = bowler_wickets.iloc[0]


# ==========================================
# 4. PLAYER OF THE MATCH
# ==========================================

pom_counts = (
    df["player_of_match"]
    .dropna()
    .value_counts()
)

top_pom_player = pom_counts.index[0]
top_pom_awards = pom_counts.iloc[0]


# ==========================================
# 5. TOSS IMPACT
# ==========================================

toss_data = df[
    df["toss_winner"].notna()
    & df["winner"].notna()
]

toss_wins_match = (
    toss_data["toss_winner"] == toss_data["winner"]
).sum()

toss_win_percentage = (
    toss_wins_match / len(toss_data)
) * 100


# ==========================================
# 6. TOP VENUE
# ==========================================

venue_counts = df["venue"].dropna().value_counts()

top_venue = venue_counts.index[0]
top_venue_matches = venue_counts.iloc[0]


# ==========================================
# 7. TOP SEASON
# ==========================================

season_counts = df["season"].value_counts()

top_season = season_counts.index[0]
top_season_matches = season_counts.iloc[0]


# ==========================================
# 8. FINAL INSIGHTS TABLE
# ==========================================

final_insights = pd.DataFrame({
    "Analysis": [
        "Team Performance",
        "Top Batsman",
        "Top Bowler",
        "Player of the Match",
        "Toss Impact",
        "Top Venue",
        "Top Season"
    ],

    "Key Finding": [
        f"{top_team} - {top_team_wins} wins",
        f"{top_batsman} - {top_batsman_runs} runs",
        f"{top_bowler} - {top_bowler_wickets} wickets",
        f"{top_pom_player} - {top_pom_awards} awards",
        f"{toss_win_percentage:.2f}% toss winners also won",
        f"{top_venue} - {top_venue_matches} matches",
        f"{top_season} - {top_season_matches} matches"
    ]
})

final_insights
```
